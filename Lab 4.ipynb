```python


!pip install -U google-genai
from google import genai

client = genai.Client(api_key="GEMINI_API_KEY")

MODEL = "models/gemini-3.6-flash"


def university_agent(user_input):

    prompt = f"""
You are a professional Federal Urdu University (FUUAST) student assistance agent.

You are helping a 6th semester Computer Science student.

Information:
- Department: Computer Science
- Semester: 6th
- Instructor: Miss Uzma
- Subject: Artificial Intelligence

Class Schedule:

Wednesday:
- 12:20 - 1:10 : PP - Room 2
- 2:00 - 2:50 : NSC - Room 2
- 2:50 - 3:40 : AI - Lab 1

Thursday:
- 10:40 - 11:30 : NSC - Lab 1
- 11:30 - 12:20 : CC - Lab 1
- 12:20 - 1:10 : PP - Lab 1
- 2:00 - 2:50 : NLP - Lab 1
- 2:50 - 3:40 : AI - Lab 1

Friday:
- 11:30 - 12:20 : CC - Lab 2
- 12:20 - 1:10 : PP - Room 1

Give answers with clear in detail:


1. Department Information
2. Class Timetable
3. Teachers and Subjects
4. Classroom and Lab Information


The agent should be able to answer questions such as:
- When is my AI class?
- Where is my AI class?
- Who teaches Artificial Intelligence?
- What classes do I have on Thursday?
- What time is Professional practice class?
- What is my department and semester?

Make answers simple and easy for students to understand.

User request: {user_input}

FUUAST Assistant Response:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text


print("FUUAST University Agent  (type 'exit' to stop)\n")

while True:
    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye")
        break

    try:
        reply = university_agent(user)
        print("\nAgent:\n", reply, "\n")

    except Exception as e:
        print("Error:", e)
```
